```python

import json
import uuid
from datetime import datetime, timezone

# Parameter controlled from the notebook or Job
dbutils.widgets.dropdown(
    "event_status",
    "IN_TRANSIT",
    ["IN_TRANSIT", "DELIVERED"],
    "Shipment status"
)

status = dbutils.widgets.get("event_status")

SOURCE_DIR = (
    "/Volumes/workspace/bronze/"
    "logistic_landing/tracking_events"
)

SHIPMENT_ID = "SHIP-DEMO-002"

# Unique filename and event ID for each execution
now = datetime.now(timezone.utc)
timestamp = now.strftime("%Y%m%dT%H%M%S%fZ")

filename = f"batch_demo_{timestamp}_{status.lower()}.json"

event = {
    "event_id": f"EVT-DEMO-{uuid.uuid4().hex}",
    "shipment_id": SHIPMENT_ID,
    "event_timestamp": now.isoformat(timespec="seconds"),
    "event_type": status,
    "distribution_center": "DC01",
    "source_system": "synthetic_tracking_api"
}

filepath = f"{SOURCE_DIR}/{filename}"

dbutils.fs.put(
    filepath,
    json.dumps(event) + "\n",
    overwrite=False
)

print(f"Created: {filepath}")
print(f"Shipment: {SHIPMENT_ID}")
print(f"Status: {status}")
```
